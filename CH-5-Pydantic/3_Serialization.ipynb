```python
from pydantic import BaseModel, Field, EmailStr
from typing import Optional, Literal

class employee_schema(BaseModel):
    id: int = Field(..., minimum=3, maximum=10, description="Employee ID must be between 3 and 10")
    age: int = Field(..., gt=18, lt=65, description="Employee age must be between 18 and 65")
    email: EmailStr = Field(..., description="Employee email must be a valid email address")
    department: Optional[str] = None
    gender: Literal["male", "female", "other"] = Field(..., description="Employee gender must be either male, female, or other")
    

obj = employee_schema(id=5, age=30, email="satyaraj21@gmail.com", department="Engineering", gender="male")
obj

```

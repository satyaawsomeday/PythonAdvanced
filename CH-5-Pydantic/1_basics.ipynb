```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    id: int
    name: str
    department: str
```

```python
obj = employee_schema(id=1, name="Satya", department="Claude")
print(obj)
```

# Coercion VS Strict Types

# Coercion

```python
obj2 = employee_schema(**{"id": "123", "name": "John", "department": "Langchain"})
obj2
```

# Strict

```python
from pydantic import StrictInt

class employee_schema(BaseModel):
    id: StrictInt
    name: str
    department: str

obj3 = employee_schema(**{"id": 123, "name": "John", "department": "Langchain"})
obj3
```
